```python
credits_df = spark.read.format("delta").load(f"{bronze_folder_path}/credits")
```

```python
cast_schema = StructType([
  StructField("cast_id", IntegerType(), True),
  StructField("character", StringType(), True),
  StructField("credit_id", StringType(), True),
  StructField("gender", IntegerType(), True),
  StructField("id", IntegerType(), True),
  StructField("name", StringType(), True),
  StructField("order", IntegerType(), True),
  StructField("profile_path", StringType(), True),
])

crew_schema = StructType([
  StructField("credit_id", StringType(), True),
  StructField("department", StringType(), True),
  StructField("gender", IntegerType(), True),
  StructField("id", IntegerType(), True),
  StructField("job", StringType(), True),
  StructField("name", StringType(), True),
  StructField("profile_path", StringType(), True)
])
```

```python
cast_df = explode_nested_json_array_silver(
    credits_df, "cast", "id", cast_schema, ["id", "name", "character", "order"]
)

crew_df = explode_nested_json_array_silver(
    credits_df, "crew", "id", crew_schema, ["id", "name", "job", "department"]
)
```

```python
cast_df.write.format("delta").save(f"{silver_folder_path}/cast")
crew_df.write.format("delta").save(f"{silver_folder_path}/crew")
```
